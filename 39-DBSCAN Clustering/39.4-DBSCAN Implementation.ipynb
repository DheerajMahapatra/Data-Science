```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN



# 1. CREATE DATASET
X, y = make_moons(n_samples=500,noise=0.08,random_state=42)

# Add some outliers
outliers = np.array([
    [1.5, 1.5],
    [1.8, 1.7],
    [-1.5, -0.5],
    [-1.7, -0.7],
    [0.5, 1.8]
])

X = np.vstack([X, outliers])


# 2. SCALE DATA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)



# 3. K-MEANS
kmeans = KMeans(n_clusters=2,random_state=42,n_init=10)

kmeans_labels = kmeans.fit_predict(X_scaled)



# 4. HIERARCHICAL CLUSTERING
hierarchical = AgglomerativeClustering(n_clusters=2,metric='euclidean',linkage="ward")

hierarchical_labels = hierarchical.fit_predict(X_scaled)



# 5. DBSCAN
dbscan = DBSCAN(eps=0.25,min_samples=5)

dbscan_labels = dbscan.fit_predict(X_scaled)



# 6. VISUALIZATION
plt.figure(figsize=(20, 5))


# BEFORE
plt.subplot(1, 4, 1)

plt.scatter(X_scaled[:, 0],X_scaled[:, 1])

plt.title("Before Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")


# K-MEANS
plt.subplot(1, 4, 2)

plt.scatter(X_scaled[:, 0],X_scaled[:, 1],c=kmeans_labels)

plt.title("K-Means")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")


# HIERARCHICAL
plt.subplot(1, 4, 3)

plt.scatter(X_scaled[:, 0],X_scaled[:, 1],c=hierarchical_labels)

plt.title("Hierarchical Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")


# DBSCAN
plt.subplot(1, 4, 4)

plt.scatter(X_scaled[:, 0],X_scaled[:, 1],c=dbscan_labels)

plt.title("DBSCAN")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")


plt.tight_layout()
plt.show()
```
