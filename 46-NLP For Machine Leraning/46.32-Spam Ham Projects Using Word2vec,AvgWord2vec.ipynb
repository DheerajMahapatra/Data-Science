````python
import pandas as pd
messages = pd.read_csv('SpamClassifier-master/smsspamcollection/SMSSpamClassifier.txt',
                    sep='\t',names=["label","message"])
````

````python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
````

````python
import re
import nltk
nltk.download('stopwords')
````

````python
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.lower()
    review = review.split()
    review = [lemmatizer.lemmatize(word) for word in review]
    review = ' '.join(review)
    corpus.append(review)
````

````python
def avg_word2vec(doc):
    # Remove out-of-vocabulary words
    vectors = [model.wv[word] for word in doc
                if word in model.wv.index_to_key]

    # Handle empty documents
    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    # Calculate average for each dimension
    return np.mean(vectors, axis=0)
````

# Pandas – `ignore_index` Explained

## 1. What is `ignore_index`?

`ignore_index` is a parameter used in Pandas functions like `pd.concat()`.

It decides whether to keep the original index labels or create a new continuous index.

- `ignore_index=False` → Keeps the original indexes.
- `ignore_index=True` → Creates a new index starting from `0`.

## 2. Syntax

```python
pd.concat([df1, df2], ignore_index=True)
```

## 3. Example Without `ignore_index=True`

```python
import pandas as pd

df1 = pd.DataFrame({"Name": ["Amit", "Rahul"]})
df2 = pd.DataFrame({"Name": ["Rohan", "Priya"]})

result = pd.concat([df1, df2])

print(result)
```

**Output:**

```text
    Name
0   Amit
1  Rahul
0  Rohan
1  Priya
```

**Explanation:**

- Pandas keeps the original indexes.
- Both DataFrames have indexes `0` and `1`.
- Therefore, duplicate indexes appear.

## 4. Example With `ignore_index=True`

```python
result = pd.concat([df1, df2], ignore_index=True)

print(result)
```

**Output:**

```text
    Name
0   Amit
1  Rahul
2  Rohan
3  Priya
```

**Explanation:**

- Pandas ignores the original row indexes.
- It creates new indexes automatically.
- The indexes start from `0` and continue sequentially.

## 5. Difference Between True and False

| Feature | False | True |
|---|---|---|
| Original index | Preserved | Ignored |
| New index | No | Yes |
| Duplicate indexes | Possible | Avoided |
| Starts from 0 | Not necessarily | Yes |

## 6. Practical Example Using a Loop

Suppose we want to combine multiple rows from an array `X` into a DataFrame.

```python
import pandas as pd
from tqdm import tqdm

df = pd.DataFrame()

for i in tqdm(range(len(X))):
    df = pd.concat(
        [df, pd.DataFrame(X[i].reshape(1, -1))],
        ignore_index=True
    )
```

**Why use `ignore_index=True` here?**

Every time a new row is added, Pandas creates a fresh continuous index instead of keeping the original index.

## 7. What Happens When `ignore_index=True` Is Not Used?

If multiple DataFrames have the same indexes, the final DataFrame may contain duplicate index labels.

This can make indexing and data handling confusing.

## 8. Important: What About `axis`?

In `pd.concat()`, `axis` decides the direction of concatenation.

- `axis=0` → Combines DataFrames row-wise.
- `axis=1` → Combines DataFrames column-wise.

**Example:**

```python
pd.concat([df1, df2], axis=0, ignore_index=True)
```

This combines rows and creates a fresh row index.

**Remember:** When `axis=1`, `ignore_index=True` resets the column labels instead of the row labels.

## 9. Quick Revision

```python
# Keep original indexes
pd.concat([df1, df2], ignore_index=False)

# Create new continuous indexes
pd.concat([df1, df2], ignore_index=True)
```

### Final Takeaway

**`ignore_index=True` means ignore the old index labels and create a new continuous index starting from 0.**

It is especially useful when combining multiple DataFrames row-wise.
