
# Difference Between Stemming and Lemmatization in NLP

In Natural Language Processing (NLP), **stemming and lemmatization** are techniques used to reduce words to their root or base form.

## 1. Stemming

- Stemming removes prefixes or suffixes from a word.
- It uses simple rules to shorten words.
- It may produce words that are not grammatically correct.
- It is faster but less accurate than lemmatization.

**Examples:**

| Original Word | Stemmed Word |
|---|---|
| Studies | studi |
| Playing | play |
| Running | run |
| Cared | care |
| Connection | connect |

## 2. Lemmatization

- Lemmatization converts a word into its meaningful dictionary base form, called a *lemma*.
- It uses vocabulary and linguistic analysis.
- It generally produces valid words.
- It is usually more accurate but slower than stemming.

**Examples:**

| Original Word | Lemmatized Word |
|---|---|
| Studies | study |
| Playing | playing |
| Running | running |
| Cared | cared |
| Better | good |

**Note:** Lemmatization depends on the part of speech (POS). For example, `playing` becomes `play` when treated as a verb, and `better` may become `good` when treated as a comparative adjective.

## 3. Difference Between Stemming and Lemmatization

| Feature | Stemming | Lemmatization |
|---|---|---|
| Technique | Removes word endings | Finds the dictionary base form |
| Output | May be an invalid word | Usually a valid word |
| Method | Rule-based chopping | Vocabulary and linguistic analysis |
| Speed | Faster | Comparatively slower |
| Accuracy | Generally lower | Generally higher |
| Context | Usually ignores context | Can use grammatical information |
| Example | Studies → studi | Studies → study |

## 4. Python Example Using NLTK

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

word = "studies"

print("Stemming:", stemmer.stem(word))
print("Lemmatization:", lemmatizer.lemmatize(word))
```

**Output:**

```text
Stemming: studi
Lemmatization: study
```

For verbs, specify the part of speech:

```python
print(lemmatizer.lemmatize("playing", pos="v"))
```

Output:

```text
play
```

## 5. When to Use Them?

- **Stemming:** Useful when speed is important, such as in basic search engines and text-processing tasks.
- **Lemmatization:** Useful when meaningful words and linguistic accuracy are important, such as in chatbots, text analysis, and NLP applications.

## Conclusion

**Stemming = Cutting word endings.**

**Lemmatization = Finding the meaningful dictionary base form.**

Both techniques help normalize words and reduce vocabulary size in NLP. Stemming is generally faster, while lemmatization usually provides more meaningful results.
