Content 5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset. Columns

* asin - ID of the product, like B000FA64PK

* reviewText - text of the review (heading).
* reviewTime - time of the review (raw).
* reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
* reviewerName - name of the reviewer.
* summary - summary of the review (description).
* unixReviewTime - unix timestamp.

````python
# Load stopwords only once
stop_words = set(stopwords.words('english'))

# Handle missing values
df['reviewText'] = df['reviewText'].fillna('').astype(str)

# Remove special characters
df['reviewText'] = df['reviewText'].apply(
    lambda x: re.sub('[^a-z A-Z 0-9]+', ' ', x)
)

# Remove the stopwords
df['reviewText'] = df['reviewText'].apply(
    lambda x: " ".join([y for y in x.split() if y not in stop_words])
)

# Remove URL
df['reviewText'] = df['reviewText'].apply(
    lambda x: re.sub(r'(http|https|ftp|ssh)://\S+|www\.\S+', '', str(x))
)

# Remove HTML tags
df['reviewText'] = df['reviewText'].apply(
    lambda x: BeautifulSoup(x, 'lxml').get_text()
)

# Remove any additional spaces
df['reviewText'] = df['reviewText'].apply(
    lambda x: " ".join(x.split())
)
````

## Word2Vec Model (added)

**Why Word2Vec?**
BOW and TF-IDF treat every word as a separate column. They don't know that *good* and *great* are similar, and they create a huge sparse matrix (one column per vocabulary word). Word2Vec learns a small dense vector (e.g. 100 numbers) for every word so that words with similar meaning get similar vectors.

**Steps**
1. Tokenize the cleaned reviews (split into list of words)
2. Train a Word2Vec model on the **train data only** (no data leakage)
3. Convert each review into one vector = average of its word vectors (Average Word2Vec)
4. Train ML classifiers and compare with BOW and TF-IDF

### Step 1: Tokenization
Word2Vec needs each review as a list of words. Our text is already cleaned and lemmatized, so a simple `split()` is enough.

````python
# Each review -> list of words
X_train_tokens = [text.split() for text in X_train]
X_test_tokens  = [text.split() for text in X_test]

print(X_train_tokens[0][:15])
print("Total train reviews:", len(X_train_tokens))
````

### Step 2: Train Word2Vec model
Important: the model is trained only on **X_train** so the test data stays unseen.

````python
from gensim.models import Word2Vec

w2v_model = Word2Vec(
    sentences=X_train_tokens,
    vector_size=100,   # size of each word vector
    window=5,          # context words considered on each side
    min_count=2,       # ignore words that appear less than 2 times
    sg=1,              # 1 = Skip-gram (better for small data), 0 = CBOW
    workers=4,
    epochs=10,
    seed=42
)

print("Vocabulary size:", len(w2v_model.wv))
````

### Step 3: Check what the model learned (sanity check)

````python
# Words with similar meaning should come close
for w in ['good', 'bad', 'story']:
    if w in w2v_model.wv:
        print(w, "->", w2v_model.wv.most_similar(w, topn=5))
        print()
````

### Step 4: Convert each review to a vector (Average Word2Vec)
For each review we take the vectors of all its words and average them. If no word of a review is in the vocabulary, we use a zero vector.

````python
import numpy as np

def avg_word2vec(tokens, model):
    vectors = [model.wv[word] for word in tokens if word in model.wv]
    if len(vectors) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vectors, axis=0)

X_train_w2v = np.array([avg_word2vec(t, w2v_model) for t in X_train_tokens])
X_test_w2v  = np.array([avg_word2vec(t, w2v_model) for t in X_test_tokens])

print("Train shape:", X_train_w2v.shape)
print("Test shape :", X_test_w2v.shape)
````

### Step 5: Train ML models on Word2Vec features
We use the same `GaussianNB` as BOW/TF-IDF for a fair comparison, and also Logistic Regression and Random Forest (they usually work better on dense vectors).

````python
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

nb_model_w2v = GaussianNB().fit(X_train_w2v, y_train)
lr_model_w2v = LogisticRegression(max_iter=1000).fit(X_train_w2v, y_train)
rf_model_w2v = RandomForestClassifier(n_estimators=200, n_jobs=-1, random_state=42).fit(X_train_w2v, y_train)
````

### Step 6: Evaluation

````python
y_pred_nb_w2v = nb_model_w2v.predict(X_test_w2v)
y_pred_lr_w2v = lr_model_w2v.predict(X_test_w2v)
y_pred_rf_w2v = rf_model_w2v.predict(X_test_w2v)

print("Word2Vec + GaussianNB accuracy        :", accuracy_score(y_test, y_pred_nb_w2v))
print("Word2Vec + Logistic Regression accuracy:", accuracy_score(y_test, y_pred_lr_w2v))
print("Word2Vec + Random Forest accuracy      :", accuracy_score(y_test, y_pred_rf_w2v))
````

````python
print("Confusion matrix (Word2Vec + Logistic Regression)")
print(confusion_matrix(y_test, y_pred_lr_w2v))
print()
print(classification_report(y_test, y_pred_lr_w2v))
````

### Step 7: Final comparison - BOW vs TF-IDF vs Word2Vec

````python
results = pd.DataFrame({
    'Model': ['BOW + GaussianNB',
                'TFIDF + GaussianNB',
                'Word2Vec + GaussianNB',
                'Word2Vec + Logistic Regression',
                'Word2Vec + Random Forest'],
    'Accuracy': [accuracy_score(y_test, y_pred_bow),
                accuracy_score(y_test, y_pred_tfidf),
                accuracy_score(y_test, y_pred_nb_w2v),
                accuracy_score(y_test, y_pred_lr_w2v),
                accuracy_score(y_test, y_pred_rf_w2v)]
})
results.sort_values('Accuracy', ascending=False).reset_index(drop=True)
````

### Conclusion
- BOW and TF-IDF create very high-dimensional sparse features, and `GaussianNB` (which assumes a normal distribution of each feature) is a poor fit for them, so accuracy stays low.
- Word2Vec gives a compact 100-dimensional dense vector that captures word meaning/context, so the classifiers get better features.
- Logistic Regression / Random Forest on Word2Vec features give the best accuracy among the models above.
- Limitation: averaging word vectors loses word order. Further improvement is possible with TF-IDF-weighted Word2Vec, pre-trained vectors (GloVe / Google News), or LSTM/BERT models.
