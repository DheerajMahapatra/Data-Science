```python
corpus = """Hello Welcome, to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
corpus
```

```python
print(corpus)
```

```python

import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
## Tokenization
## Sentence-->Paragraphs
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentences in documents:
    print(sentences)
```

```python
## Tokenization
## Paragraph-->Words
## Sentence-->Words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentences in documents:
    print(word_tokenize(sentences))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordDetokenizer
```

```python
tokenizer = TreebankWordDetokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

# Difference Between NLTK and spaCy in NLP

| Basis | NLTK | spaCy |
|---|---|---|
| Full Form | Natural Language Toolkit | No official full form |
| Purpose | Mainly used for learning and research in NLP | Mainly used for real-world NLP applications |
| Speed | Generally slower | Generally faster |
| Tokenization | Supports word and sentence tokenization | Supports fast word and sentence tokenization |
| Stemming | Supports stemming | Does not provide a built-in stemmer as a standard pipeline component |
| Lemmatization | Supports lemmatization | Supports lemmatization |
| POS Tagging | Available | Available |
| Named Entity Recognition (NER) | Available, often requiring separate setup | Built-in pretrained pipelines are available |
| Ease of Learning | Good for beginners and understanding NLP concepts | Suitable for practical NLP tasks |
| Applications | NLP experiments, education, and research | Chatbots, information extraction, and production applications |

## Conclusion

NLTK is suitable for learning NLP concepts and conducting research, whereas spaCy is generally preferred for building fast, production-ready NLP applications.
